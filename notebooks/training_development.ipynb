## Preprocessing and Feature Engineering

```python
df["weekly_study_hours_actual_missing"] = df["weekly_study_hours_actual"].isna().astype(int)
df["weekly_study_hours_actual"] = df.groupby("plan_type")["weekly_study_hours_actual"].transform(lambda x: x.fillna(x.median()))
```

```python
df.dropna(subset=["mentor_contact_freq_per_month"], inplace=True)
df.dropna(subset=["message_response_time_hours"], inplace=True)
```

```python
df.isnull().sum()
```

```python
df["satisfaction_missing"] = df["satisfaction_survey_score"].isna().astype(int)
```

```python
df["satisfaction_survey_score"] = (
    df.groupby("plan_type")["satisfaction_survey_score"]
      .transform(lambda x: x.fillna(x.median()))
)
```

calling preprocessed data Again for Engineering

```python
data = pd.read_csv("../data/updated_data.csv")
```

```python
data.shape
```

```python
from config import FEATURES , STUDENT_INFO , TARGET_FEATURE , CAT_COLS
```

```python
X = data.drop(columns=STUDENT_INFO)
X_last = X.drop(columns=TARGET_FEATURE)
cat_features = [X_last.columns.get_loc(c) for c in CAT_COLS]
y = data[TARGET_FEATURE]

```

```python
X_train , X_test , y_train , y_test = train_test_split(
    X_last,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
for col in CAT_COLS:
    X_train[col] = X_train[col].astype(str)
    X_test[col] = X_test[col].astype(str)
```

## Model Building for Ensembled learning

```python
from catboost import CatBoostClassifier
```

```python
model = CatBoostClassifier(
    iterations=300,
    depth=4,
    learning_rate=0.05,
    cat_features=cat_features,
    auto_class_weights="Balanced",
    eval_metric="AUC",
    random_state=42,
    verbose=False,
)

model.fit(
    X_train, y_train,
    eval_set=(X_test, y_test),
    early_stopping_rounds=50,
)
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score, classification_report

proba = model.predict_proba(X_test)[:, 1]
pred = (proba >= 0.5).astype(int)

print("ROC-AUC:", roc_auc_score(y_test, proba))
print("PR-AUC (Average Precision):", average_precision_score(y_test, proba))
print(classification_report(y_test, pred))
```
